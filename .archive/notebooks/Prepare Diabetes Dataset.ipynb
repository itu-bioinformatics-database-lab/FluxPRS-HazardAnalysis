### Delete PRS That have less than 20 scores

```python
import os
import shutil
import pandas as pd

base_path = "../DIABETES/Final_Prs_Scores"

list_dirs = [
    d for d in os.listdir(base_path)
    if d != ".ipynb" and os.path.isdir(os.path.join(base_path, d))
]

for pgs in list_dirs:
    csv_path = os.path.join(base_path, pgs, "prs_scores.csv")

    if not os.path.exists(csv_path):
        continue

    df = pd.read_csv(csv_path)
    n = df.shape[0]

    if n < 20:
        folder_path = os.path.join(base_path, pgs)
        print(f"Deleting folder: {folder_path}")
        shutil.rmtree(folder_path)

# Refresh list after deletion
list_dirs = [
    d for d in os.listdir(base_path)
    if d != ".ipynb" and os.path.isdir(os.path.join(base_path, d))
]

print("Remaining folders:", list_dirs)
```
