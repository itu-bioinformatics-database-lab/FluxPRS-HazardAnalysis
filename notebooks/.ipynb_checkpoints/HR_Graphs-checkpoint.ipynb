```python
import pandas as pd
import matplotlib.pyplot as plt
import numpy as np
from matplotlib.lines import Line2D

def rename_model(model_name):
    if model_name == "Metabolite Levels Pathway Combined PRS":
        return "Fluxomics (Pathway-based) + Metabolomics + PRS"
    elif model_name == "Metabolite Levels Pathway Combined":
        return "Fluxomics (Pathway-based) + Metabolomics (No PRS)"
    elif model_name == "Metabolite Levels PRS":
        return "Metabolomics + PRS (No Fluxomics)"
    elif model_name == "Reaction Diff Scores PRS":
        return "Fluxomics (Reaction-based) + PRS"
    elif model_name == "Pathway Diff Scores PRS":
        return "Fluxomics (Pathway-based) + PRS"
    elif model_name == "Prs ScorePGS003957":
        return "PRS"
    elif model_name == "Metabolite Levels Reaction Combined PRS":
        return "Fluxomics (Reaction-based) + Metabolomics + PRS"
    elif model_name == "Metabolite Levels Reaction Combined":
        return "Fluxomics (Reaction-based) + Metabolomics (No PRS)"
    elif model_name == "Reaction Diff Scores":
        return "Fluxomics (Reaction-based) (No PRS)"
    elif model_name == "Pathway Diff Scores":
        return "Fluxomics (Pathway-based) (No PRS)"
    elif model_name == "Metabolite Levels":
        return "Metabolomics (No PRS, No Fluxomics)"
    else:
        return model_name

fluxomics_models = {
    "Fluxomics (Pathway-based) + Metabolomics + PRS",
    "Fluxomics (Pathway-based) + Metabolomics (No PRS)",
    "Fluxomics (Reaction-based) + PRS",
    "Fluxomics (Pathway-based) + PRS",
    "Fluxomics (Reaction-based) + Metabolomics + PRS",
    "Fluxomics (Reaction-based) + Metabolomics (No PRS)",
    "Fluxomics (Reaction-based) (No PRS)",
    "Fluxomics (Pathway-based) (No PRS)"
}

df = pd.read_csv("../ROSMAP-AD/processed/hazard_ratios.csv")
df.rename(columns={df.columns[0]: "PRS"}, inplace=True)
df.set_index("PRS", inplace=True)
df.columns = [rename_model(c) for c in df.columns]
df = df.apply(pd.to_numeric, errors='coerce')
df.rename(columns={df.columns[-1]: "PRS"}, inplace=True)

model_names = list(df.columns)
num_colors = len(model_names) - 1
colors = plt.cm.get_cmap('tab10', num_colors)

plt.figure(figsize=(14, 10))
y_pos = np.arange(len(df.index))

def get_marker(model_name):
    if model_name in fluxomics_models:
        return 's'  # kare
    else:
        return 'o'  # yuvarlak

def get_color(model_name):
    if model_name == "PRS":
        return 'red'
    elif model_name == "Fluxomics (Pathway-based) + PRS":
        return 'gold'
    else:
        prs_idx = model_names.index("PRS")
        idx = model_names.index(model_name)
        if idx > prs_idx:
            idx -= 1
        return colors(idx)

for i, col in enumerate(model_names):
    plt.scatter(df[col], y_pos + i * 0.1,
                label=col,
                marker=get_marker(col),
                color=get_color(col),
                s=50, alpha=0.8)

plt.yticks(y_pos + 0.4, df.index)
plt.xlabel('Hazard Ratio')
plt.title('Hazard Ratios Across Different Models per PRS (Forest Plot Style)')
plt.grid(axis='x', linestyle='--', alpha=0.7)

# Renk legend'i (grafikteki renkler)
handles_color = [
    Line2D([0], [0], marker='o', color='w', label=name,
           markerfacecolor=get_color(name), markersize=8)
    for name in model_names
]

legend1 = plt.legend(
    handles=handles_color,
    title="Models (Colors)",
    loc='lower center',
    bbox_to_anchor=(0.5, -0.35),
    ncol=4,
    fontsize='small',
    title_fontsize='small',
    frameon=False
)

plt.gca().add_artist(legend1)

# Marker shape legend'i (legend'deki markerlar Gri renkte)
handles_shape = [
    Line2D([0], [0], marker='o', color='w',
           label='No Fluxomics (Only PRS and/or Metabolomics)',
           markerfacecolor='gray', markersize=8),
    Line2D([0], [0], marker='s', color='w',
           label='Fluxomics-Inclusive Models',
           markerfacecolor='gray', markersize=8),
]

legend2 = plt.legend(
    handles=handles_shape,
    title="Marker Shapes",
    loc='lower center',
    bbox_to_anchor=(0.5, -0.50),
    ncol=1,
    fontsize='small',
    title_fontsize='small',
    frameon=False
)

plt.tight_layout()
plt.savefig("hazard_ratio_forest_plot_300-ROSMAP-AD.png", dpi=300, bbox_inches='tight')
plt.show()
```

```python
import pandas as pd
import matplotlib.pyplot as plt
import numpy as np
from matplotlib.lines import Line2D
import warnings
warnings.filterwarnings("ignore")

def rename_model(model_name):
    mapping = {
        "Metabolite Levels Pathway Combined PRS": "Fluxomics (Pathway-based) + Metabolomics + PRS",
        "Metabolite Levels Pathway Combined": "Fluxomics (Pathway-based) + Metabolomics (No PRS)",
        "Metabolite Levels PRS": "Metabolomics + PRS (No Fluxomics)",
        "Reaction Diff Scores PRS": "Fluxomics (Reaction-based) + PRS",
        "Pathway Diff Scores PRS": "Fluxomics (Pathway-based) + PRS",
        "Prs ScorePGS003957": "PRS",
        "Metabolite Levels Reaction Combined PRS": "Fluxomics (Reaction-based) + Metabolomics + PRS",
        "Metabolite Levels Reaction Combined": "Fluxomics (Reaction-based) + Metabolomics (No PRS)",
        "Reaction Diff Scores": "Fluxomics (Reaction-based) (No PRS)",
        "Pathway Diff Scores": "Fluxomics (Pathway-based) (No PRS)",
        "Metabolite Levels": "Metabolomics (No PRS, No Fluxomics)",
    }
    return mapping.get(model_name, model_name)

fluxomics_models = {
    "Fluxomics (Pathway-based) + Metabolomics + PRS",
    "Fluxomics (Pathway-based) + Metabolomics (No PRS)",
    "Fluxomics (Reaction-based) + PRS",
    "Fluxomics (Pathway-based) + PRS",
    "Fluxomics (Reaction-based) + Metabolomics + PRS",
    "Fluxomics (Reaction-based) + Metabolomics (No PRS)",
    "Fluxomics (Reaction-based) (No PRS)",
    "Fluxomics (Pathway-based) (No PRS)",
}

def get_marker(model_name):
    return "s" if model_name in fluxomics_models else "o"

def build_plot(
    csv_path="../DIABETES/processed/hazard_ratios.csv",
    out_path="hazard_ratio_forest_plot_300-DIABETES-test.png",
    figsize=(14, 10),
    use_log_x=False,            # set True if you want log-scale
    dodge=True,                 # True = separate models vertically per PRS row (recommended)
    dodge_span=0.26,            # total up/down span for dodging
    add_hr1_line=True,
    point_size=50,
    alpha=0.85,
):
    # --- Load ---
    df = pd.read_csv(csv_path)

    # Ensure first column is PRS id
    df.rename(columns={df.columns[0]: "PRS"}, inplace=True)
    df.set_index("PRS", inplace=True)

    # Rename model columns
    df.columns = [rename_model(c) for c in df.columns]

    # Coerce to numeric
    df = df.apply(pd.to_numeric, errors="coerce")

    # If you have a duplicate/last PRS column coming from file, rename it safely (optional)
    # This line in your original code was risky; keep only if you know the last column is PRS values.
    # df.rename(columns={df.columns[-1]: "PRS"}, inplace=True)

    model_names = list(df.columns)

    # --- Colors ---
    # Keep PRS red, Pathway+PRS gold, others from tab10
    other_models = [m for m in model_names if m not in ("PRS", "Fluxomics (Pathway-based) + PRS")]
    cmap = plt.cm.get_cmap("tab10", max(1, len(other_models)))
    color_map = {}

    for i, m in enumerate(other_models):
        color_map[m] = cmap(i)

    color_map["PRS"] = "red"
    color_map["Fluxomics (Pathway-based) + PRS"] = "gold"

    def get_color(model_name):
        return color_map.get(model_name, "black")

    # --- Y positions ---
    y_pos = np.arange(len(df.index))

    if dodge:
        offsets = np.linspace(-dodge_span, dodge_span, len(model_names))
    else:
        offsets = np.zeros(len(model_names))

    # --- Plot ---
    plt.figure(figsize=figsize)

    for j, col in enumerate(model_names):
        x = df[col].values
        y = y_pos + offsets[j]

        plt.scatter(
            x,
            y,
            label=col,
            marker=get_marker(col),
            color=get_color(col),
            s=point_size,
            alpha=alpha,
        )

    # Center y ticks on the PRS rows (no +0.4 trick)
    plt.yticks(y_pos, df.index)
    plt.xlabel("Hazard Ratio")
    plt.title("Hazard Ratios Across Different Models per PRS (Forest Plot Style)")
    plt.grid(axis="x", linestyle="--", alpha=0.7)

    if add_hr1_line:
        plt.axvline(1.0, color="black", linestyle="--", linewidth=1, alpha=0.8)

    if use_log_x:
        # Only safe if all HR values are > 0
        plt.xscale("log")

    # --- Legends ---
    # 1) Color legend (models)
    handles_color = [
        Line2D(
            [0],
            [0],
            marker="o",
            color="w",
            label=name,
            markerfacecolor=get_color(name),
            markersize=8,
        )
        for name in model_names
    ]

    legend1 = plt.legend(
        handles=handles_color,
        title="Models (Colors)",
        loc="lower center",
        bbox_to_anchor=(0.5, -0.35),
        ncol=4,
        fontsize="small",
        title_fontsize="small",
        frameon=False,
    )
    plt.gca().add_artist(legend1)

    # 2) Shape legend
    handles_shape = [
        Line2D(
            [0],
            [0],
            marker="o",
            color="w",
            label="No Fluxomics (Only PRS and/or Metabolomics)",
            markerfacecolor="gray",
            markersize=8,
        ),
        Line2D(
            [0],
            [0],
            marker="s",
            color="w",
            label="Fluxomics-Inclusive Models",
            markerfacecolor="gray",
            markersize=8,
        ),
    ]

    plt.legend(
        handles=handles_shape,
        title="Marker Shapes",
        loc="lower center",
        bbox_to_anchor=(0.5, -0.50),
        ncol=1,
        fontsize="small",
        title_fontsize="small",
        frameon=False,
    )

    plt.tight_layout()
    plt.savefig(out_path, dpi=300, bbox_inches="tight")
    plt.show()
```
