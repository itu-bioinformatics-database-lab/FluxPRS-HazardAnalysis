```python
import sys, os
from pathlib import Path
import warnings 
warnings.filterwarnings("ignore")

# --- robust project root detection ---
CWD = Path.cwd()
PROJECT_ROOT = next((p for p in [CWD] + list(CWD.parents) if (p / "preprocessing").exists()), None)
if PROJECT_ROOT is None:
    raise RuntimeError("Could not find PROJECT_ROOT (no 'preprocessing/' directory found in parents).")

# Make imports work + ensure relative paths are consistent
sys.path.insert(0, str(PROJECT_ROOT))
os.chdir(PROJECT_ROOT)

print("CWD:", CWD)
print("PROJECT_ROOT:", PROJECT_ROOT)

import pandas as pd
import seaborn as sns
import matplotlib.pyplot as plt
import matplotlib.patches as mpatches
from sklearn.preprocessing import StandardScaler
import os
main_folder = 'ROSMAP-AD/Final_Prs_Scores'
for subfolder in os.listdir(main_folder):
    if '.DS_Store' == subfolder:
        continue
    prs_scores = pd.read_csv(f"{main_folder}/{subfolder}/prs_scores.csv")
    rosmap_clinical = pd.read_csv('ROSMAP-AD/raw/clinical.csv')     # Contains ID, Study, and projid

    # Merge on individual ID
    rosmap_clinical['sample_id'] = rosmap_clinical['Sample ID'].astype(str)
    merged = prs_scores.merge(rosmap_clinical, on='sample_id', how='inner')
    merged = merged[['sample_id', 'score']]

    metabolon_270 = pd.read_csv("ROSMAP-AD/raw/metabolite_levels.csv")
    second_merged = pd.merge(merged, metabolon_270, left_on='sample_id', right_on='Sample ID', how='inner')

    second_merged = second_merged[['sample_id', 'score', 'sample_id', 'Diagnosis']]
    scaler = StandardScaler()
    second_merged["score_z"] = scaler.fit_transform(second_merged[["score"]])
    # Set plot style
    sns.set(style="whitegrid")
    palette = {"AD": "blue", "Control": "red"}
    # Create the density plot
    plt.figure(figsize=(10, 6))
    sns.kdeplot(data=second_merged, x="score_z", hue="Diagnosis", fill=False, common_norm=False, palette=palette, alpha=0.5)

    legend_handles = [
        mpatches.Patch(color=palette["AD"], label="AD"),
        mpatches.Patch(color=palette["Control"], label="Control")
    ]
    # Plot labels and title
    plt.title(f"Density Distribution of {subfolder} Scores by Diagnosis")
    plt.xlabel("PRS Score")
    plt.ylabel("Density")

    plt.legend(handles=legend_handles, title="Diagnosis")
    plt.tight_layout()
    plt.show()
```

```python
from analysis.config import make_default_paths
# dataset root
DATA_DIR = PROJECT_ROOT / "ROSMAP-AD"
paths = make_default_paths(DATA_DIR)

print("PROJECT_ROOT:", PROJECT_ROOT)
print("DATA_DIR:", DATA_DIR)
metabolite_levels_file_path = paths.metabolite_levels_file_path
pathway_file_path = paths.pathway_file_path
x_pathway_path = paths.x_pathway_path
```

```python
import pandas as pd
import seaborn as sns
import matplotlib.pyplot as plt
from analysis import analyse_with_logistic_regression
from preprocessing import PathwayPreprocessorFactory

# Get data
factory = PathwayPreprocessorFactory(metabolite_levels_file_path, pathway_file_path, x_pathway_path)
preprocessor = factory.create_preprocessor()
y, risk_scores, sample_ids_ = analyse_with_logistic_regression(preprocessor)

# Create DataFrame
df = pd.DataFrame({
    'Diagnosis': y.map({0: "Control", 1: "AD"}),  # Map directly here
    'score': risk_scores
})

# Set plot style
sns.set(style="whitegrid")
palette = {"AD": "blue", "Control": "red"}

# Create the density plot
plt.figure(figsize=(10, 6))
sns.kdeplot(data=df, x="score", hue="Diagnosis", fill=False, common_norm=False, palette=palette, alpha=0.5)

# Plot labels and title
plt.title("Density Distribution of PRS Scores by Diagnosis")
plt.xlabel("PRS Score")
plt.ylabel("Density")
plt.legend(title="Diagnosis")
plt.tight_layout()
plt.show()
```
